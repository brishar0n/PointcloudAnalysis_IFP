```python
import laspy
import numpy as np
import open3d as o3d
# import pandas as pd
import seaborn as sns
import open3d as o3d
import open3d.visualization.gui as gui
import open3d.visualization.rendering as rendering
import sys
```

```python
file=laspy.read(r'C:\Users\messa\Menu\my_coding\rmit\Programming Project\dev_loader\PointcloudAnalysis_IFP\preprocessed\bologna\low_featured.laz')
x,y,z=file.x,file.y,file.z
```

```python
features=[x for x in file.point_format.dimension_names]
```

```python
labels=np.array(file.classification).astype(int).transpose()
```

```python
points=np.column_stack((x,y,z))
```

```python
pcd=o3d.geometry.PointCloud()
pcd.points=o3d.utility.Vector3dVector(points)
```

```python
np.unique_counts(labels)
```

- 0 not yet classified (nothing done yet)
- 1 unclassified (actively marked as nothing)
- 2 ground, sidewalk
- 3,4,5 vegetation, low(gras) medium(shrubbery) high (trees)
- 6 buildings
- 8 street furniture
- 10 street markings
- 11 street, pavement
- 12 bike lanes
- 13 temporary things(bicycles, trashcans)
- 15 cars, trucks

```python
palette = {
    0:  [0.5, 0.5, 0.5], # Not yet classified
    2:  [0.7, 0.7, 0.8], # Sidewalk
    5:  [0.0, 0.4, 0.0], # High Veg
    8:  [1.0, 0.9, 0.0], # Furniture
    11: [0.2, 0.2, 0.2], # Street
    13: [1.0, 0.4, 0.7], # Temp
    15: [1.0, 0.6, 0.0], # Vehicles
}
```

```python
colors=np.array([palette.get(l) for l in labels])
```

```python
pcd.colors=o3d.utility.Vector3dVector(colors)
```

```python
o3d.visualization.draw_geometries([pcd])
```

```python
import open3d.visualization.gui as gui
import open3d.visualization.rendering as rendering
```

```python
app=gui.Application.instance
app.initialize()
win=app.create_window('testing',1920,1080)
view=gui.SceneWidget()
view.scene=rendering.Open3DScene(win.renderer)
view.scene.add_geometry('pcd',pcd,rendering.MaterialRecord())
win.add_child(view)
app.run()
```
